```python
from tkinter import *
from tkinter import ttk

win = Tk()

label = Label(win, text='python',font=30,fg="blue")
label.pack()

separator = ttk.Separator(win,orient="horizontal")
separator.pack(fill=X)

label_1 = Label(win, text='java',font=30,fg="pink")
label_1.pack()


win.mainloop()
```

# vertical

```python
from tkinter import *
from tkinter import ttk

win = Tk()

label = Label(win, text='python',font=30,fg="blue")
label.pack(side=RIGHT)

separator = ttk.Separator(win,orient="vertical")
separator.pack(fill=Y,side=RIGHT)

label_1 = Label(win, text='java',font=30,fg="pink")
label_1.pack(side=RIGHT)


win.mainloop()
```
