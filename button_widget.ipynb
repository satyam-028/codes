# background color

# relief parameter

# image in button

# compound

```python
from tkinter import *

win = Tk()

win.config(bg="black")
photo = PhotoImage(file="C:/Users/ssaty/Downloads/dog.png")
bt = Button(win ,text="ON",font=(30),bg="green",relief="flat",image=photo,compound="left")
bt.place(x=100,y=100)

win.mainloop()
```

 # command paramter

```python
from tkinter import *
def python():
    lb.config(text="python")

              
win = Tk()

win.config(bg="black")
bt = Button(win ,text="ON",font=(30),bg="green",command=python)
bt.place(x=100,y=100)


lb = Label(win, text= "hello",font=(30))
lb.place(x=200,y=200)



win.mainloop()
```
