# change the cursor/arrow

```python
from tkinter import *

win = Tk()

win.config(bg="red")
lab = Label(win, text="hello! welcome to delhi",font=("Arial Rounded MT Bold",30,"bold"),bg="red",
          cursor= "star") # bg remove box gary color
lab.place(x=100,y=200)

win.mainloop()
```

# relief parameter

```python
from tkinter import *

win = Tk()

win.config(bg="red")
lab = Label(win, text="hello! welcome to delhi",font=("Arial Rounded MT Bold",30,"bold"),bg="pink"
          ,relief="sunken") # bg remove box gary color
lab.place(x=100,y=200)

win.mainloop()
```

# justify parameter

```python
from tkinter import *

win = Tk()

win.config(bg="red")
lab = Label(win, text="hello! welcome\n to \ndelhi",font=("Arial Rounded MT Bold",30,"bold"),bg="pink"
          ,relief="sunken",justify=LEFT) # bg remove box gary color
lab.place(x=100,y=200)

win.mainloop()
```

# text variables

```python
from tkinter import *

win = Tk()

win.config(bg="red")

var = StringVar()

lab = Label(win, font=("Arial Rounded MT Bold",30,"bold"),bg="pink",textvariable=var
          ,relief="sunken",justify=LEFT) # bg remove box gary color

lab.place(x=100,y=200)

var.set("Hello! satyam welcome to delhi")

win.mainloop()
```

# user input 

```python
from tkinter import *

entry = input("Enter your data")
win = Tk()

win.config(bg="red")

var = StringVar()

lab = Label(win, font=("Arial Rounded MT Bold",30,"bold"),bg="pink",textvariable=var
          ,relief="sunken",justify=LEFT) # bg remove box gary color

lab.place(x=100,y=200)

var.set(entry)

win.mainloop() 
```

# under line on text

```python
from tkinter import *

win = Tk()

win.config(bg="red")
lab = Label(win, text="hello! welcome\n to \ndelhi",font=("Arial Rounded MT Bold",30,"bold"),bg="pink"
          ,relief="sunken",justify=LEFT,underline=4) # bg remove box gary color
lab.place(x=100,y=200)

win.mainloop()
```
