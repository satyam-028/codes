# Strip plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
var = sns.load_dataset("tips")
var
```

```python
sns.stripplot(var)
plt.show()
```

```python
sns.stripplot(x="day",y="total_bill",data=var)
plt.show()
```

# hue parameter

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex")
plt.show()
```

# palette parameter

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r")
plt.show()
```

# linewidth

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r",linewidth=1)
plt.show()
```

# edgecolor

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r",linewidth=1,edgecolor="r")
plt.show()
```

# jitter parameter

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r",linewidth=1,edgecolor="r",jitter=3)
plt.show()
```

# size parameter

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r",linewidth=1,edgecolor="r",size=10)
plt.show()
```

# marker parameter

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r",edgecolor="r",marker="<")
plt.show()
```

# alpha parameter

```python
sns.stripplot(x="day",y="total_bill",data=var,hue="sex",palette="rocket_r",edgecolor="r",alpha=0.2)
plt.show()
```

# single plot

```python
sns.stripplot(x=var["total_bill"],palette="rocket_r",edgecolor="r",alpha=0.2)
plt.show()
```
