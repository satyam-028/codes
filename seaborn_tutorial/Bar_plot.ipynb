# Bar plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("penguins")
df
```

```python
sns.barplot(x=df.island,y=df.bill_depth_mm)
plt.show()
```

```python
sns.barplot(x="island",y="bill_depth_mm",data=df)
plt.show()
```

# hue parameter

```python
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex")
plt.show()
```

# order parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order)
plt.show()
```

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,hue_order=["Female","Male"])
plt.show()
```

# ci parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,hue_order=["Female","Male"],ci=100)
plt.show()
```

# n_boot parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,ci=60,n_boot=2)
plt.show()
```

# orient parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v")
plt.show()
```

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="bill_length_mm",y="bill_depth_mm",data=df,orient="h")
plt.show()
```

# color parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,color="black")
plt.show()
```

# palette parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v",palette="icefire")
plt.show()
```

# saturation parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v",palette="icefire",saturation=0.3)
plt.show()
```

# errcolor parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v",palette="icefire",saturation=0.3,errcolor="pink")
plt.show()
```

# errwidth parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v",palette="icefire",saturation=0.3,errcolor="pink",errwidth=7)
plt.show()
```

# capsize parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v",palette="icefire",saturation=0.3,errcolor="pink",errwidth=3,capsize=0.1)
plt.show()
```

# dodge parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v",palette="icefire",saturation=0.3,errcolor="pink",errwidth=3,dodge=False)
plt.show()
```

# alpha parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,orient="v",palette="icefire",saturation=100,errcolor="pink",errwidth=3,alpha=0.7)
plt.show()
```
