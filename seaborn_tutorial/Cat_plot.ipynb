# Cat plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
var = sns.load_dataset("tips")
var
```

```python
sns.catplot(x="tip",y="size",data=var)
plt.show()
```

```python
sns.catplot(x="size",y="tip",data=var,kind="point")
plt.show()
```

```python
sns.catplot(x="size",y="tip",data=var,kind="box")
plt.show()
```

```python
sns.catplot(x="size",y="tip",data=var,kind="bar")
plt.show()
```

```python
sns.catplot(x="size",y="tip",data=var,kind="swarm")
plt.show()
```

```python
sns.catplot(x="size",y="tip",data=var,kind="violin")
plt.show()
```

```python
sns.catplot(x="size",y="tip",data=var,kind="boxen")
plt.show()
```

```python
sns.catplot(x="size",y="tip",data=var,kind="strip")
plt.show()
```
