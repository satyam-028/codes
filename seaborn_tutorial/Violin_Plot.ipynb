# Violin plot

```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = sns.load_dataset("tips")
data
```

```python
sns.violinplot(x="day",y="total_bill",data=data)
plt.show()
```

# hue parameter

```python
sns.violinplot(x="day",y="total_bill",data=data,hue="time")
plt.show()
```

# linewidth parameter

```python
sns.violinplot(x="day",y="total_bill",data=data,linewidth=6)
plt.show()
```

# palette parameter

```python
sns.violinplot(x="day",y="total_bill",data=data,palette="Dark2")
plt.show()
```

# order parameter

```python
sns.violinplot(x="time",y="total_bill",data=data,order=["Dinner","Lunch"])
plt.show()
```

# saturation parameter

```python
sns.violinplot(x="time",y="total_bill",data=data,palette="Dark2",saturation=0.2)
plt.show()
```

# color parameter

```python
sns.violinplot(x="day",y="total_bill",data=data,color="black")
plt.show()
```

# split parameter

```python
sns.violinplot(x="day",y="total_bill",data=data,hue="sex",split=True)
plt.show()
```

# scale parameter

```python
# sns.violinplot(x="day",y="total_bill",data=data,hue="sex",split=True,scale="count")
# sns.violinplot(x="day",y="total_bill",data=data,hue="sex",split=True,scale="width")
sns.violinplot(x="day",y="total_bill",data=data,hue="sex",split=True,scale="area")
plt.show()
```

```python
# sns.violinplot(x="day",y="total_bill",data=data,hue="sex",split=True,scale="count")
sns.violinplot(x="day",y="total_bill",data=data,hue="sex",split=True,scale="width")
plt.show()
```

```python
sns.violinplot(x="day",y="total_bill",data=data,hue="sex",split=True,scale="count")
plt.show()
```
