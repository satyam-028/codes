# Pair plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
var = sns.load_dataset("tips")
var
```

```python
sns.pairplot(var)
plt.show()
```

```python
var
```

# hue parameter

```python
sns.pairplot(var,hue="sex")
plt.show()
```

# vars parameter

```python
sns.pairplot(var,vars=["total_bill","tip"],hue="sex")
plt.show()
```

# hue_order parameter

```python
sns.pairplot(var,vars=["total_bill","tip"],hue="sex",hue_order=['Female','Male'])
plt.show()
```

# palette parameter

```python
sns.pairplot(var,vars=["total_bill","tip"],hue="sex",hue_order=['Female','Male'],palette="BuGn")
plt.show()
```

# x_vars

```python
sns.pairplot(var,hue="sex",hue_order=['Female','Male'],x_vars=["total_bill","tip"])
plt.show()
```

# y_vars parameter

```python
sns.pairplot(var,hue="sex",hue_order=['Female','Male'],y_vars=["total_bill","tip"])
plt.show()
```

# kind parameter

```python
sns.pairplot(var,hue="sex",hue_order=['Female','Male'],kind="kde") #'scatter','kde','hist','reg'
plt.show()
```

# diag_kind

```python
sns.pairplot(var,hue="sex",hue_order=['Female','Male'],kind="kde",diag_kind='hist') #'scatter','kde','hist','reg'
plt.show()
```

# markers parameter

```python
sns.pairplot(var,hue="sex",hue_order=['Female','Male'],markers=["*",">"])
plt.show()
```
