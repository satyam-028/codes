# Histogram plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

var = sns.load_dataset("penguins")
var
```

```python
sns.displot(var["bill_length_mm"])
plt.show()
```

# bins parameter

```python
sns.displot(var["flipper_length_mm"],bins=[170,180,190,200,210,220,230,240])
plt.show()
```

# kde parameter

```python
sns.displot(var["flipper_length_mm"],kde=True)
plt.show()
```

# rug parameter

```python
sns.displot(var["flipper_length_mm"],kde=True,rug=True)
plt.show()
```

# color parameter

```python
sns.displot(var["flipper_length_mm"],kde=True,rug=True,color="pink")
plt.show()
```

# log_scale parameter

```python
sns.displot(var["flipper_length_mm"],kde=True,rug=True,color="pink",log_scale=True)
plt.show()
```
