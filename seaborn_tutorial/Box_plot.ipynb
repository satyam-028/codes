# Box plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
var = sns.load_dataset("tips")
var
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var)
plt.show()
```

# hue parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="time")
plt.show()
```

# color parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",color="b")
plt.show()
```

# order parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",color="b",order=["Fri","Sun","Thur","Sat"])
plt.show()
```

# showmeans parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",color="b",order=["Fri","Sun","Thur","Sat"],showmeans=True)
plt.show()
```

# meanprops parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",color="black",order=["Fri","Sun","Thur","Sat"],showmeans=True,meanprops={"marker":"+","markeredgecolor":"b"})
plt.show()
```

# linewidth parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",color="black",order=["Fri","Sun","Thur","Sat"],showmeans=True,meanprops={"marker":"+","markeredgecolor":"b"},linewidth=2)
plt.show()
```

# palette parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",palette="nipy_spectral_r",order=["Fri","Sun","Thur","Sat"],showmeans=True,meanprops={"marker":"+","markeredgecolor":"b"},linewidth=2)
plt.show()
```

# orient parameter

```python
sns.set(style="whitegrid")
sns.boxplot(x="day",y="total_bill",data=var,hue="sex",palette="nipy_spectral_r",order=["Fri","Sun","Thur","Sat"],showmeans=True,meanprops={"marker":"+","markeredgecolor":"b"},linewidth=2,orient="v")
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(x="total_bill",y="day",data=var,hue="sex",palette="nipy_spectral_r",order=["Fri","Sun","Thur","Sat"],showmeans=True,meanprops={"marker":"+","markeredgecolor":"b"},linewidth=2,orient="h")
plt.show()
```

```python
sns.set(style="whitegrid")
sns.boxplot(data=var,palette="nipy_spectral_r",showmeans=True,meanprops={"marker":"+","markeredgecolor":"b"},linewidth=2,orient="h")
plt.show()
```
