```python
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd
import numpy as np


```

```python
var = np.linspace(1,10,20).reshape(4,5)
var
```

```python
sns.heatmap(var)
```

```python
data = sns.load_dataset("anagrams")
x = data.drop(columns=["attnr"],axis=1).head(10)
x
```

```python
sns.heatmap(x)
```

# mini and max parameter

```python
sns.heatmap(var,vmin=0,vmax=12)
```

# cmap parameter

```python
sns.heatmap(var,vmin=0,vmax=10,cmap="PuOr")
```

# annot parameter

```python
sns.heatmap(var,vmin=0,vmax=10,cmap="PuOr",annot=True)
```

```python
var_1 = np.linspace(1,10,10).reshape(2,5)
var_1
```

```python
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True)
```

```python
ar = np.array([["a0","a1","a2","a3","a4"],["b0","b1","b2","b3","b4"]])
ar
```

```python
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=ar,fmt="s")
```

```python
y = {"fontsize":20,"color":"r"}
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y)
```

# linewidth parameter

```python
y = {"fontsize":20,"color":"r"}
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10)
```

# linecolor parameter

```python
y = {"fontsize":20,"color":"r"}
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10,linecolor="y")
```

# cbar parameter

```python
y = {"fontsize":20,"color":"r"}
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10,linecolor="y",cbar=False)
```

```python
y = {"fontsize":20,"color":"r"}
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10,linecolor="y",cbar=False)
plt.show()
```

# xticklabels parameter

```python
y = {"fontsize":20,"color":"r"}
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10,linecolor="y",cbar=False,xticklabels=False)
plt.show()
```

# yticklabels parameter

```python
y = {"fontsize":20,"color":"r"}
sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10,linecolor="y",cbar=False,yticklabels=False)
plt.show()
```

# labels parameter

```python
y = {"fontsize":20,"color":"r"}
v = sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10,linecolor="y",cbar=False,yticklabels=False)
v.set(xlabel="python",ylabel="data")
plt.show()
```

# font_scale parameter

```python
y = {"fontsize":20,"color":"r"}
v = sns.heatmap(var_1,vmin=0,vmax=10,cmap="PuOr",annot=True,annot_kws=y,linewidth=10,linecolor="y",cbar=False,yticklabels=False)
v.set(xlabel="python",ylabel="data")
sns.set(font_scale=10)
plt.show()
```
