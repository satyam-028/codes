# Scatter plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = sns.load_dataset("penguins").head(100)
df
```

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df)
plt.show()
```

# hue parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex")
plt.show()
```

# style parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex")
plt.show()
```

# size parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",size="sex")
plt.show()
```

# sizes parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",size="sex",sizes=(100,40))
plt.show()
```

# palette parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",size="sex",sizes=(100,40),palette="gist_ncar")
plt.show()
```

# alpha parameter

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",size="sex",sizes=(100,40),palette="gist_ncar",alpha=0.3)
plt.show()
```

# marker parameter

```python
m = {"Male":"*","Female":"o"}
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df,hue="sex",style="sex",size="sex",sizes=(100,40),palette="gist_ncar",markers=m)
plt.show()
```
