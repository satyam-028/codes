# Line Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

var = [1,2,3,4,5,6,7]
var_1 = [3,4,2,5,3,8,9]

x_1 = pd.DataFrame({"var":var, "var_1":var_1})

sns.lineplot(x="var",y="var_1",data=x_1)

# plt.plot(var,var_1)
plt.show()
```

# Seabron Dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

y_1 = sns.load_dataset("penguins")

var = [1,2,3,4,5,6,7]
var_1 = [3,4,2,5,3,8,9]

x_1 = pd.DataFrame({"var":var, "var_1":var_1})

sns.lineplot(x="var",y="var_1",data=x_1)

plt.show()
```

```python
y_1 = sns.load_dataset("penguins")
y_1
```

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=y_1)

plt.show()
```

# hue parameter

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=y_1,hue='sex')

plt.show()
```

# size parameter

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=y_1,hue='sex',size=50)

plt.show()
```

# style parameter

```python
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=y_1,hue='sex',size=50,style="sex")

plt.show()
```
