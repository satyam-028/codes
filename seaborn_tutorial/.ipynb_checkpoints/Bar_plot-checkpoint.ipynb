# Bar plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("penguins")
df
```

```python
sns.barplot(x=df.island,y=df.bill_depth_mm)
plt.show()
```

```python
sns.barplot(x="island",y="bill_depth_mm",data=df)
plt.show()
```

# hue parameter

```python
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex")
plt.show()
```

# order parameter

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order)
plt.show()
```

```python
order = ["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_depth_mm",data=df,hue="sex",order=order,hue_order=["Female","Male"])
plt.show()
```
