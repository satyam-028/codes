# styling plot

```python
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
var = sns.load_dataset("tips")
var
```

```python
sns.boxplot(y=var["total_bill"])
plt.show()
```

```python
sns.barplot(x=var["day"],y=var['total_bill'])
plt.show()
```

```python
sns.barplot(x=var["day"],y=var['total_bill'])
plt.grid()
plt.show()
```

```python
sns.set_style("dark")
sns.barplot(x=var["day"],y=var['total_bill'])
plt.show()
```

```python
sns.set_style("whitegrid")
sns.barplot(x=var["day"],y=var['total_bill'])
plt.show()
```

```python
sns.set_style("darkgrid")
sns.barplot(x=var["day"],y=var['total_bill'])
plt.show()
```

```python
sns.set_style("ticks")
sns.barplot(x=var["day"],y=var['total_bill'])
plt.show()
```

```python
sns.set_style("darkgrid")
sns.barplot(x=var["day"],y=var['total_bill'])
plt.grid()
plt.show()
```

```python
sns.set_style("white")
sns.barplot(x=var["day"],y=var['total_bill'])
sns.despine()
plt.show()
```

```python
sns.set_style("white")
plt.figure(figsize=(12,3))
sns.barplot(x=var["day"],y=var['total_bill'])
sns.despine()

plt.show()
```

```python
sns.set_style("white")
plt.figure(figsize=(3,12))
sns.barplot(x=var["day"],y=var['total_bill'])
sns.despine()

plt.show()
```

```python
sns.set_style("white")
sns.set_context("poster",font_scale=0.7)
sns.barplot(x=var["day"],y=var['total_bill'])
sns.despine()

plt.show()
```

```python
sns.set_style("white")
sns.set_context("paper",font_scale=0.7)
sns.barplot(x=var["day"],y=var['total_bill'])
sns.despine()

plt.show()
```

```python
sns.set_style("white")
sns.set_context("paper",font_scale=0.7)
sns.barplot(x=var["day"],y=var['total_bill'],palette="PiYG_r")
sns.despine()

plt.show()
```
