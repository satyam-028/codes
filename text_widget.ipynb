# button 

```python
from tkinter import *

win = Tk()

def python():
    var = text.get("1.0","end")
    label.config(text=var)

text = Text(win, font=("Times New Roman",19),bg="red",fg="black")
text.place(x=10,y=10,width=600,height=100)

label = Label(win, text="",font=("Times New Roman",19))
label.place(x=10,y=200)

button = Button(win, text="ok",font=("Times New Roman",19),command=python)
button.place(x=10,y=130,width=60,height=50)

win.mainloop()
```
