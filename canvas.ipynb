# line in canvas

```python
from tkinter import *

win = Tk()

canvas_1 = Canvas(win,bg="red")
canvas_1.place(x=0,y=0,height=700,width=700)

coordinate1 = 0,0,700,700
coordinate2 = 700,0,0,700

line1 = canvas_1.create_line(coordinate1,fill="Blue")
line2 = canvas_1.create_line(coordinate2,fill="yellow")

win.mainloop()
```

# arc in canvas

```python
from tkinter import *

win = Tk()

canvas_1 = Canvas(win,bg="red")
canvas_1.place(x=0,y=0,height=700,width=700)

coordinate = 20,20,200,200

arc = canvas_1.create_arc(coordinate,start=0,extent=90,fill= "Blue")

win.mainloop()
```

 # oval in canvas

```python
from tkinter import *

win = Tk()

canvas = Canvas(win,bg="red")
canvas.place(x=0,y=0,height=700,width=700)

oval = canvas.create_oval(20,20,200,300,fill="yellow")

win.mainloop()
```

# polygon in canvas

```python
from tkinter import *

win = Tk()

canvas = Canvas(win,bg="red")
canvas.place(x=0,y=0,height=700,width=700)

coordinate = [20,20,50,50,40,100]

canvas.create_polygon(coordinate,fill="yellow")


win.mainloop()
```
