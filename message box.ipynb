```python
from tkinter import *
from tkinter.messagebox import showinfo,showerror,showwarning

win = Tk()
def test():
    showinfo(title="python info",message= "hello python")
    
button = Button(win,text="ok",command=test)
button.place(x=10,y=10,height=30,width=40)

win.mainloop()
```

# ask ok_cancel function

```python
from tkinter import *
from tkinter.messagebox import askokcancel

win = Tk()
def okcancel():
     ans = askokcancel(title="python info",message= "hello python")
     print(ans)
    
button = Button(win,text="ok",command=okcancel)
button.place(x=10,y=10,height=30,width=40)

win.mainloop()
```

# ask retry_cancel function

```python
from tkinter import *
from tkinter.messagebox import askretrycancel

win = Tk()
def retrycancel():
    ans = askretrycancel(title="python info",message= "hello python")
    if ans:
        retrycancel()
    else:
        label.config(text="thanks you")
 
button = Button(win,text="ok",command=retrycancel)
button.place(x=10,y=10,height=30,width=40)


label = Label(win,text="")
label.place(x=10,y=100)

win.mainloop()
```
