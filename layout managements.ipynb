```python
from tkinter import *

win = Tk()

win.title("data")
win['bg'] = "yellow"

lab = Label(win, text = "satyam classes",font=("Time New Roman",30,"bold"),bg = "red")
lab.pack(padx=100,pady=100) ## padx and ypad work on external space
# lab.pack(ipadx=20,ipady=20) ## ipadx and ipady work on internal text 

win.mainloop()
```

# fill parameter on pack

```python
from tkinter import *

win = Tk()

win.title("data")
win['bg'] = "yellow"

lab = Label(win, text = "satyam classes",font=("Time New Roman",30,"bold"),bg = "red")
lab.pack(ipadx=20,fill="x") 
# lab.pack(ipady=20,fill="y") 

win.mainloop()
```

# side parameter

```python
from tkinter import *

win = Tk()

win.title("data")
win['bg'] = "yellow"

lab = Label(win, text = "satyam",font=("Time New Roman",30,"bold"),bg = "red")
lab.pack(padx=20,pady=20,side = "left")

lab1 = Label(win, text = "classes",font=("Time New Roman",30,"bold"),bg = "red")
lab1.pack(padx=20,pady=20,side = "left")

win.mainloop()
```

# grid() method 

```python
from tkinter import *

win = Tk()

win.title("data")
win['bg'] = "yellow"

lab = Label(win, text = "satyam",font=("Time New Roman",30,"bold"),bg = "red")
lab.grid(row=0,column=0)

lab1 = Label(win, text = "classes",font=("Time New Roman",30,"bold"),bg = "red")
lab1.grid(row=1,column=0)


win.mainloop()
```

# place() method
