```python
from tkinter import *

def value_change(var):
    var = value.get()
    lb.config(text=var)


win = Tk()

op_list = ["c++","python","java","c"]

value = StringVar()

value.set("option")

op = OptionMenu(win,value,*op_list,command=value_change)
op.place(x=100,y=100,height=30,width=100)

lb = Label(win, text="")
lb.place(x=300,y=100,height=30,width=100)


win.mainloop()
```
