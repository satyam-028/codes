# Core Data Structures in Pandas

Series

```python
import pandas as pd

s = pd.Series([1,2,3,455,4,6,8])
print(s)
type(s)
```

```python
s2 = pd.Series([1,2,3,455,4,6,8], index = ['sam','satyam','iram','shaba','aayush','sid','jon'])
print(s2)
```

```python
s2['sam']
```

```python
s2['iram']
```
