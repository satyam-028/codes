# Box_plot

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70]

plt.boxplot(x)
plt.show()
```

# Box shape notch parameter

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70]

plt.boxplot(x,notch=True)

plt.show()
```

# vertical and horziontial Box

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70]

plt.boxplot(x,vert=False)

plt.show()
```

# width of box

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70]

plt.boxplot(x,widths=0.1)

plt.show()
```

# label

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70]

plt.boxplot(x,labels=["data"])


plt.show()
```

# patch artist

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70]

plt.boxplot(x,labels=["data"],patch_artist=True)


plt.show()
```

# show mean

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70]

plt.boxplot(x,labels=["data"],showmeans=True)
 
plt.show()
```

# whis parameter

# sym 

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70,120]

plt.boxplot(x,labels=["data"],showmeans=True,sym="g+")
# plt.boxplot(x,labels=["data"],showmeans=True,sym=" ")
 
plt.show()
```

# box props parameter

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70,120]

plt.boxplot(x,labels=["data"],showmeans=True,sym="g+",boxprops=dict(color="r"))
 
plt.show()
```

# cap color box

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70,120]

plt.boxplot(x,labels=["data"],showmeans=True,sym="g+",capprops=dict(color="g"))
 
plt.show()
```

# whisker props

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70,120]

plt.boxplot(x,labels=["data"],showmeans=True,sym="g+",whiskerprops=dict(color="r"))
 
plt.show()
```

# flier props

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70,120]



plt.boxplot(x,labels=["data"],showmeans=True,sym="g+",whiskerprops=dict(color="r"),
           flierprops=dict(markeredgecolor="y"))
 
plt.show()
```

# mutipler plot

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70,120]
x1 = [10,20,40,50,30,60,90]


plt.boxplot(x,labels=["data"],showmeans=True,sym="g+",whiskerprops=dict(color="r"),
           flierprops=dict(markeredgecolor="y"))
plt.boxplot(x1,labels=["data"],showmeans=True,sym="g+",whiskerprops=dict(color="r"),
           flierprops=dict(markeredgecolor="y"))
 
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [10,20,30,40,50,60,70,120]
x1 = [10,20,40,50,30,60,90]

y = [x,x1]
plt.boxplot(y,labels=["python","c++"],showmeans=True,sym="g+",whiskerprops=dict(color="r"),
           flierprops=dict(markeredgecolor="y"))
 
plt.show()
```
