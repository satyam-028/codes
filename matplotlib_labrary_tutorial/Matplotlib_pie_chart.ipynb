# Matplotlib_pie_plot

```python
import matplotlib.pyplot as plt

x = [10,20,30,40]
y = ["C","C++","Java","Python"]
ex = [0.4,0.0,0.2,0.0]

plt.pie(x,labels=y,explode=ex)

plt.show()
```

## color on graph

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
y = ["C","C++","Java","Python"]
ex = [0.4,0.0,0.0,0.0]
c = ["r","b","c","m"]

plt.pie(x,labels=y,explode=ex,colors=c)

plt.show()
```

# Percentage on graph

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
y = ["C","C++","Java","Python"]
ex = [0.4,0.0,0.0,0.0]
c = ["r","b","c","m"]

plt.pie(x,labels=y,explode=ex,colors=c,autopct="%0.1f%%") # %0.1 means one zero after point and 0.2 two zero after point

plt.show()
```
