# title on graph

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,5,2,4]

plt.plot(x,y)

plt.title("data",fontsize=15)

plt.show()
```

 # label on graph

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,5,2,4]

plt.plot(x,y)

plt.title("data",fontsize=15)
plt.xlabel("days",fontsize=15)
plt.ylabel("python",fontsize=15)

plt.show()
```

# text 

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,4,2,5]

plt.plot(x,y)

plt.title("data",fontsize=15)
plt.xlabel("days",fontsize=15)
plt.ylabel("python",fontsize=15)

plt.text(3,4,"pyhton",fontsize=15)

plt.show()
```

# style

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,4,2,5]

plt.plot(x,y)

plt.title("data",fontsize=15)
plt.xlabel("days",fontsize=15)
plt.ylabel("python",fontsize=15)

plt.text(3,4,"pyhton",fontsize=15,style="italic")

plt.show()
```

# bbox parameter

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,4,2,5]

plt.plot(x,y)

plt.title("data",fontsize=15)
plt.xlabel("days",fontsize=15)
plt.ylabel("python",fontsize=15)

plt.text(3,4,"pyhton",fontsize=15,style="italic",bbox={"facecolor":"m"})

plt.show()
```

# annotate parameter

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,4,2,5]

plt.plot(x,y)

plt.title("data",fontsize=15)
plt.xlabel("days",fontsize=15)
plt.ylabel("python",fontsize=15)

plt.text(3,4,"pyhton",fontsize=15,style="italic",bbox={"facecolor":"m"})
plt.annotate("python",xy=(3,2),xytext=(4,4),arrowprops=dict(facecolor="black",shrink=100))

plt.show()
```

# legend

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,4,2,5]

plt.plot(x,y)

plt.title("data",fontsize=15)
plt.xlabel("days",fontsize=15)
plt.ylabel("python",fontsize=15)

# plt.text(3,4,"pyhton",fontsize=15,style="italic",bbox={"facecolor":"m"})
# plt.annotate("python",xy=(3,2),xytext=(4,4),arrowprops=dict(facecolor="black",shrink=100

plt.legend(["up"],loc=9,facecolor="r",edgecolor="y")                                                            

plt.show()
```

# framealpha

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,4,2,5]

plt.plot(x,y)

plt.title("data",fontsize=15)
plt.xlabel("days",fontsize=15)
plt.ylabel("python",fontsize=15)

# plt.text(3,4,"pyhton",fontsize=15,style="italic",bbox={"facecolor":"m"})
# plt.annotate("python",xy=(3,2),xytext=(4,4),arrowprops=dict(facecolor="black",shrink=100

plt.legend(["up"],loc=9,facecolor="r",edgecolor="y",framealpha=0.5)                                                            

plt.show()
```
