# x ticks

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,5,2,4]

plt.plot(x,y)

plt.xticks(x)

plt.show()
```

# y ticks

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,5,2,4]

plt.plot(x,y)

plt.xticks(x)
plt.yticks(y)

plt.show()
```

# labels on x axis

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,5,2,4]

plt.plot(x,y)

plt.xticks(x,labels=["python","c","c++","java","php"])
plt.yticks(y)

plt.show()
```

# limit on axis

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,5,2,4]

plt.plot(x,y)

# plt.xticks(x,labels=["python","c","c++","java","php"])
# plt.yticks(y)

plt.xlim(0,10)
plt.ylim(0,10)

plt.show()
```

# x-y axis

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [3,1,5,2,4]

plt.plot(x,y)

plt.axis([0,10,0,7])

plt.show()
```
