# savefig parameter

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,6,4]

plt.plot(x,y,color="r")

plt.savefig("line")

plt.show()
```

# dpi parameter

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,6,4]

plt.plot(x,y,color="r")

plt.savefig("line1",dpi=2000)

plt.show()
```

# face color

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,6,4]

plt.plot(x,y,color="r")

plt.savefig("line1",dpi=2000,facecolor="y")

plt.show()
```

# formet parameter

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,6,4]

plt.plot(x,y,color="r")

plt.savefig("line2.pdf",dpi=2000,facecolor="y")
plt.savefig("line2.jpg",dpi=2000,facecolor="y")

plt.show()
```

# transparent parameter garph

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,6,4]

plt.plot(x,y,color="r")

plt.savefig("line3",dpi=2000,facecolor="y",transparent=True)
# plt.savefig("line3",dpi=2000,facecolor="y",transparent=False)

plt.show()
```

# box_inches

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,6,4]

plt.plot(x,y,color="r")

plt.savefig("line4",dpi=2000,facecolor="y",transparent=True,bbox_inches="tight")

plt.show()
```
