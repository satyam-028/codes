# step_plot

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y)

plt.show()
```

# gird parameter

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y)

plt.grid()
plt.show()
```

# title on graph

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y)


plt.title("python",fontsize=15)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

# marker on graph

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y,marker="*",)


plt.title("python",fontsize=15)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

# color on line 

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y,marker="*",color="r")


plt.title("python",fontsize=15)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

# marker size on graph

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y,marker="*",ms=10)


plt.title("python",fontsize=15)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

# marker color

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y,marker="*",ms=10,mfc="pink")


plt.title("python",fontsize=15)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

# label on graph

```python
import matplotlib.pyplot as plt
x = [1,2,6,4,3]
y = [1,2,9,2,5]

plt.step(x,y,marker="*",ms=10,mfc="pink",label="data")


plt.title("python",fontsize=15)
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.legend()
plt.legend(loc=1)
plt.grid()
plt.show()
```
