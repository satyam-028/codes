# Stem_plot

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,3,2,7,8,9]

plt.stem(x,y)

plt.show()
```

# Line style

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,3,2,7,8,9]

# plt.stem(x,y,linefmt="-.")
# plt.stem(x,y,linefmt=":")
# plt.stem(x,y,linefmt="-")
plt.stem(x,y,linefmt="--")

plt.show()
```

# line style color

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,3,2,7,8,9]

# plt.stem(x,y,linefmt="-.")
# plt.stem(x,y,linefmt=":",markerfmt="r+")
# plt.stem(x,y,linefmt="-")
plt.stem(x,y,linefmt="--",markerfmt="b+")

plt.show()
```

# bottom parameter

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,3,2,7,8,9]

plt.stem(x,y,linefmt="--",markerfmt="b+",bottom=3)

plt.show()
```

# base line color

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,3,2,7,8,9]

plt.stem(x,y,linefmt="--",markerfmt="b+",bottom=3,basefmt="g")

plt.show()
```

# label parameter

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,3,2,7,8,9]

plt.stem(x,y,linefmt="--",markerfmt="b+",bottom=3,basefmt="g",label="Data",use_line_collection=False)

plt.legend()
plt.show()
```

# orientation 

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,3,2,7,8,9]

plt.stem(x,y,linefmt="--",markerfmt="b+",bottom=3,basefmt="g",label="Data",orientation="horizontal")

plt.legend()
plt.show()
```
