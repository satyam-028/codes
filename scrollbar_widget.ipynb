```python
from tkinter import *
from tkinter import ttk

win = Tk()
text = Text(win,font=30)
text.place(x=10,y=10,height=500,width=600)

scrollbar = Scrollbar(win,orient="vertical",command=text.yview)
scrollbar.place(x=640,y=10,height=500,width=20)

win.mainloop()
```
