```python
from tkinter import *
from tkinter import ttk

def call(p):
    p = str(var.get())
    label.config(text=p)

win = Tk()

var = DoubleVar()

scale = Scale(win,from_ = 0,to = 100,fg="red",orient="vertical",variable=var,command=call)
scale.place(x=100,y=100,height=300,width=60)

label = Label(win,text="",font=50)
label.place(x=200,y=100)

win.mainloop()
```
