```python
from tkinter import *
from tkinter import ttk

win = Tk()

var = IntVar()
sp = ttk.Spinbox(win,from_ = 0,to=10,font=30,textvariable=var)
sp.place(x=10,y=10,height=30,width=300)

win.mainloop()
```

# wrap parameter

```python
from tkinter import *
from tkinter import ttk

win = Tk()

var = IntVar()
sp = ttk.Spinbox(win,from_ = 0,to=10,font=30,textvariable=var,wrap=True)
sp.place(x=10,y=10,height=30,width=300)

win.mainloop()
```

# command parameter

```python
from tkinter import *
from tkinter import ttk

def python():
    label.config(text=str(var.get()))


win = Tk()

var = IntVar()
sp = ttk.Spinbox(win,from_ = 0,to=10,font=30,textvariable=var,command=python)
sp.place(x=10,y=10,height=30,width=300)

label = Label(win,text="",font=30)
label.place(x=10,y=50)

win.mainloop()
```
