# deselect() parameter

```python
from tkinter import *

win = Tk()

rd = Radiobutton(win, text="python",value="python")
rd.deselect()
rd.pack()

rd1 = Radiobutton(win, text="java",value="java")
rd1.deselect()
rd1.pack()

win.mainloop()
```

```python
from tkinter import *
def ok():
    lb.config(text=var.get())

win = Tk()

list_1 = (("python","p"),("java","j"),("c++","+"))

var = StringVar()

for i in list_1:
    rd = Radiobutton(win, text=i[0],value=i[1],variable=var,command=ok)
    rd.pack()
    
lb = Label(win, text="", font=30)
lb.pack()

win.mainloop()
```
