```python
from tkinter import *
from tkinter import ttk

win = Tk()

def entry():
    p = var.get()
    label.config(text=p)

list_1 = ["c#","c++","python","java"]

var = StringVar()

combobox = ttk.Combobox(win,values=list_1,height=50,width=40,textvariable=var)
combobox.set("Enter")
combobox.place(x=100,y=100)

label = Label(win,text="",font=30)
label.place(x=100,y=200)

button = Button(win,text='ok',command=entry)
button.place(x=100,y=300)

win.mainloop()
```
