```python
from tkinter import *

win = Tk()

win.config(bg="red")
photo = PhotoImage(file="C:/Users/ssaty/Downloads/dog.png")

lb = Label(win, image=photo,bg="red", text="dog",compound="top", font=(30))
lb.place(x=50 ,y=50)


win.mainloop()
```

# height and width of image

```python
from tkinter import *

win = Tk()

win.config(bg="red")
photo = PhotoImage(file="C:/Users/ssaty/Downloads/dog.png")

lb = Label(win, image=photo,bg="red")
lb.place(x=100 ,y=100, height=100,width=50)


win.mainloop()
```

# label frame widget

```python
from tkinter import *

win = Tk()

lable_frame = LabelFrame(win, text= "python", font=(30))
lable_frame.place(x=100,y=100, height=100,width=500)

win.mainloop()
```

# labelanchor parameter

```python
from tkinter import *

win = Tk()

lable_frame = LabelFrame(win, text= "python", font=(30),labelanchor=N)
lable_frame.place(x=100,y=100, height=100,width=500)

win.mainloop()
```

# labelframe inside label

```python
from tkinter import *

win = Tk()

lable_frame = LabelFrame(win, text= "", font=(30),labelanchor=N)
lable_frame.place(x=100,y=100, height=100,width=500)

lable_1 = Label(win, text= "python", font=(30))
lable_1.place(x=150,y=150)

win.mainloop()
```
