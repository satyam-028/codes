# two frame function

```python
from tkinter import *
from tkinter import ttk

win = Tk()

frame_1 = Frame(win,bd=4,bg='red',relief="sunken")
frame_1.place(x=10,y=10,height=500,width=500)

frame_2 = Frame(win,bd=4,bg='yellow',relief="sunken")
frame_2.place(x=500,y=0,height=500,width=500)

win.mainloop()
```

# label function

```python
from tkinter import *
from tkinter import ttk

win = Tk()

frame_1 = Frame(win,bd=4,bg='red',relief="sunken")
frame_1.place(x=10,y=10,height=500,width=500)

label = Label(frame_1,text="hello")
label.place(x=100,y=100)


frame_2 = Frame(win,bd=4,bg='yellow',relief="sunken")
frame_2.place(x=500,y=0,height=500,width=500)

win.mainloop()
```
