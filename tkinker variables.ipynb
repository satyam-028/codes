```python
from tkinter import *

win = Tk()

var = StringVar(win, value="hello")
print(var.get())

win.mainloop()
```

# set() method

```python
from tkinter import *

win = Tk()

var = StringVar(win)
var.set("python")
print(var.get())



win.mainloop()
```

# int() method 

```python
from tkinter import *

win = Tk()

var = IntVar(win, value= 12)
var = IntVar(win, value= 14.34)
print(var.get())



win.mainloop()
```

# double() method

```python
from tkinter import *

win = Tk()

var = DoubleVar(win, value= 12)
print(var.get())



win.mainloop()
```

# boolean() method 

```python
from tkinter import *

win = Tk()

var = BooleanVar(win, value= True)
print(var.get())



win.mainloop()
```
