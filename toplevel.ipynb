```python
from tkinter import *

win = Tk()

def top():
    tp = Toplevel(win)
    tp.title("java")
    tp.config(bg="yellow")
    label = Label(tp,text="thank you")
    label.place(x=10,y=10)
    tp.mainloop()
    
win.title("python")    
win.config(bg="pink")    

button = Button(win,text="ok",command=top)
button.place(x=20,y=20,heigh=100,width=100)


win.mainloop()
```
