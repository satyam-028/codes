```python
from tkinter import *

win = Tk()

def change():
    status_bar.config(text="Go")


status_bar = Label(win,text="ready",relief=SUNKEN,anchor=W,bd=8)
status_bar.pack(side=BOTTOM,fill=X)

button = Button(win,text="ok",command=change)
button.place(x=10,y=10,height=50,width=50)

win.mainloop()
```
