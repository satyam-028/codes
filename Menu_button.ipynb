```python
from tkinter import *

win = Tk()

mb = Menubutton(win, text="file")
mb.menu =  Menu(mb)
mb["menu"] = mb.menu


mb.menu.add_command(label = "New")
mb.menu.add_command(label = "open")
mb.menu.add_command(label = "save")

mb.pack()

win.mainloop()
```

# tearoff parameter

```python
from tkinter import *

win = Tk()

mb = Menubutton(win, text="file")
mb.menu =  Menu(mb,tearoff=0)
mb["menu"] = mb.menu


mb.menu.add_command(label = "New")
mb.menu.add_command(label = "open")
mb.menu.add_command(label = "save")

mb.pack()

win.mainloop()
```

# Menu()

```python
from tkinter import *

win = Tk()

file_menu = Menu(win)
f_menu =  Menu(file_menu,tearoff=0)
# f_menu["menu"] = file_menu.menu


f_menu.add_command(label = "New")
f_menu.add_command(label = "open")
f_menu.add_command(label = "save")
f_menu.add_command(label = "save as")
f_menu.add_separator()
f_menu.add_command(label = "Exit")

win.config(menu=file_menu)
file_menu.add_cascade(label="file",menu=f_menu)

win.mainloop()
```

# multi file

```python
from tkinter import *

win = Tk()

Main_menu = Menu(win)
win.config(menu=Main_menu)


f_menu =  Menu(Main_menu,tearoff=0)
f_menu.add_command(label = "New")
f_menu.add_command(label = "open")
f_menu.add_command(label = "save")
f_menu.add_command(label = "save as")
f_menu.add_separator()
f_menu.add_command(label = "Exit")

Main_menu.add_cascade(label="file",menu=f_menu)



f_menu1 =  Menu(Main_menu,tearoff=0)
f_menu1.add_command(label = "cut")
f_menu1.add_command(label = "copy")
f_menu1.add_command(label = "paste")
f_menu1.add_command(label = "delete")
f_menu1.add_separator()
f_menu1.add_command(label = "Undo")

Main_menu.add_cascade(label="Edit",menu=f_menu1)

win.mainloop()
```

# sub menu()

```python
from tkinter import *

win = Tk()

Main_menu = Menu(win)
win.config(menu=Main_menu)


f_menu =  Menu(Main_menu,tearoff=0)
f_menu.add_command(label = "New")
f_menu.add_command(label = "open")
f_menu.add_separator()

f_menu.add_command(label = "save")
f_menu.add_separator()

sub_menu =  Menu(f_menu,tearoff=1)
sub_menu.add_command(label = "save cut")
sub_menu.add_command(label = "save file where")
f_menu.add_cascade(label="save as",menu=sub_menu)

f_menu.add_separator()
f_menu.add_command(label = "Exit")
Main_menu.add_cascade(label="file",menu=f_menu)




f_menu1 =  Menu(Main_menu,tearoff=0)
f_menu1.add_command(label = "cut")
f_menu1.add_command(label = "copy")
f_menu1.add_command(label = "paste")
f_menu1.add_command(label = "delete")
f_menu1.add_separator()
f_menu1.add_command(label = "Undo")

Main_menu.add_cascade(label="Edit",menu=f_menu1)

win.mainloop()
```
