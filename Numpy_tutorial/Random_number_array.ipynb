# rand()

```python
import numpy as np


arr = np.random.rand(4)
print(arr)
```

```python
var_1 = np.random.rand(2,5)
print(var_1)
print(type(var_1))
```

# randn()

```python
var_3 = np.random.randn(5)
print(var_3)
```

# ranf()

```python
var_4 = np.random.ranf(5)
print(var_4)
```

# randint()

```python
var_5 = np.random.randint(1,10,1)
print(var_5)
```
