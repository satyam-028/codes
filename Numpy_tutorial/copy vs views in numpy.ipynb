# copy vs view

```python
import numpy as np

var = np.array([1,2,3,4,5])
co = var.copy()
print("var :",var)
print("copy :",co)
```

```python
x = np.array([9,8,7,6,5])
vi = x.view()
print("x :",x)
print("view :",vi)
```

```python
import numpy as np

var = np.array([1,2,3,4,5])
co = var.copy()
var[1] = 9
print("var :",var)
print("copy :",co)
```

```python
x = np.array([9,8,7,6,5])
vi = x.view()
x[3] = 4
print("x :",x)
print("view :",vi)
```
