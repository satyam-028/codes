# join array

```python
import numpy as np

var = np.array([1,2,3,4])
var1 = np.array([5,6,7,8])
# arr = np.concatenate((var,var1))
print(np.concatenate((var,var1)))
```

```python
var2 = np.array([[1,2],[3,4]])
var3 = np.array([[5,6],[7,8]])
arr1 = np.concatenate((var2,var3),axis=1)
print(arr1)
```

```python
var = np.array([1,2,3,4])
var1 = np.array([5,6,7,8])
arr = np.stack((var,var1),axis=1)
print(arr)
```

```python
var = np.array([1,2,3,4])
var1 = np.array([5,6,7,8])
arr = np.hstack((var,var1))
print(arr)
```

```python
var = np.array([1,2,3,4])
var1 = np.array([5,6,7,8])
arr = np.vstack((var,var1))
print(arr)
```

```python
var = np.array([1,2,3,4])
var1 = np.array([5,6,7,8])
arr = np.dstack((var,var1)) ### d means height
print(arr)
```
