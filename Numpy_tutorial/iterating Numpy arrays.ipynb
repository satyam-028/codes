# iteration

```python
import numpy as np

var = np.array([4,6,8,4,2,6,8,9])
print(var)
print()
for i in var:
    print(i)
```

```python
var1 = np.array([[1,2,3,4],[6,7,8,9]])
print(var1)
print()
for i in var1:
    print(i)
```

```python
var1 = np.array([[1,2,3,4],[6,7,8,9]])
print(var1)
print()
for i in var1:
    print(i)
print()

for k in var1:
    for l in k:
        print(l)
```

```python
var3 = np.array([[[1,2,3,4],[5,6,7,8]]])
print(var3)

for i in var3:
    for k in i:
        for j in k:
            print(j)
```

# nditer function

```python
var3 = np.array([[[1,2,3,4],[5,6,7,8]]])
print(var3)

for i in np.nditer(var3):
    print(i)
```

```python
var3 = np.array([[[1,2,3,4],[5,6,7,8]]])
print(var3)
print(var3.ndim)
print()

for i in np.nditer(var3,flags=['buffered'],op_dtypes=["S"]):
    print(i)
```

# ndenumerate

```python
var3 = np.array([[[1,2,3,4],[5,6,7,8]]])
print(var3)
print()
for i,d in np.ndenumerate(var3):
    print(i,d)
```
