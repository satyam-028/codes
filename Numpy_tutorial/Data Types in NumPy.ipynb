```python
import numpy as np

arr = np.array([23,4,3,23])
arr.dtype
```

```python
arr = np.array([23.3,4.2,6.7,23.6])
arr.dtype
```

```python
arr = np.array(["A","C","W","F","G"])
arr.dtype
```

```python
arr = np.array(["A","C","W","F","G",2,3,4,6])
arr.dtype
```

```python
arr.astype("float64")
```

```python
arr2 = np.array([23,4,3,23], dtype = "float32")
arr2
```

```python
arr_int64 = np.array([1,2,3], dtype= np.int64)
arr_int32 = np.array([1,2,3], dtype= np.int32)

print(arr_int64.nbytes)
print(arr_int32.nbytes)
```

```python
x = np.array([1,2,3,4],dtype=np.int8)
print("Data Type :",x.dtype)
print(x)
```

```python
x_1 = np.array([1,2,3,4],dtype="f")
print("Data Type :",x_1.dtype)
print(x_1)
```

```python
x_2 = np.array([1,2,3,4])

# new = np.float_(x_2)
new = np.float32(x_2)

print("Data Type :",x_2.dtype)
print("Data Type :",new.dtype)
print(x_2)
print(new)
```

```python
x_3 = np.array([1,2,3,4])

new = np.float32(x_3)
new_1 = np.int_(new)

print("Data Type :",x_3.dtype)
print("Data Type :",new.dtype)
print("Data Type :",new_1.dtype)
print(x_3)
print(new) 
print(new_1) 
```
