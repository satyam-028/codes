# Checking Array Properties

```python
arr = np.array([[10, 20, 30], [40, 50, 60],[70,80,90],[22,34,55]])
print(arr)

print("\nShape", arr.shape)
print("\nSize", arr.size)
print("\nDimensions", arr.ndim)
print("\nData type", arr.dtype)
```

# 1D array

```python
arr = [1,2,3,4]
y = np.array(arr)
print(y)
print(type(arr))
print(y.ndim)
```

# 2D

```python
arr_1 = [[1,2,3,4],[3,4,5,6]]
y_1 = np.array(arr_1)
print(y_1)
print(type(arr_1))
print(y_1.ndim)
```

# 3D

```python
arr_2 = [[[1,2,3,4],[3,4,5,6],[4,5,6,7]]]
y_2 = np.array(arr_2)
print(y_2)
print(type(arr_2))
print(y_2.ndim)
```

# multi D array

```python
arr_3 = [1,2,3,5]
y_3 = np.array(arr_3,ndmin=10)
print(y_3)
print(type(arr_3))
print(y_3.ndim)
```
