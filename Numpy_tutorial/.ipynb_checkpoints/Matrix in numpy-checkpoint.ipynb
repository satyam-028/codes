```python
import numpy as np

var = np.matrix([[1,2,3],[4,5,6]])
print(var)
print(type(var))
```

```python
var1 = np.array([[1,2,3],[4,5,6]])
print(var1)
print(type(var1))
```

```python
var = np.matrix([[1,2,3],[4,5,6]])
var2 = np.matrix([[1,2,3],[4,5,6]])
print(var,"\n",var2)
print(type(var))
print(var+var2)
print(var-var2)
print(var*var2)
```

```python
var = np.matrix([[1,2],[4,5]])
var2 = np.matrix([[1,2],[4,5]])
print(var*var2)
```

```python
var = np.matrix([[1,2],[4,5]])
var2 = np.matrix([[1,2],[4,5]])
print(var.dot(var2))
```

# Transpose

```python
var3 = np.array([[1,2,3],[4,5,6]])
print(var1)
print()
print(np.transpose(var3))
```

```python
var0 = np.array([[1,2,3],[4,5,6]])
print(var1)
print()
print(var0.T)
```

# Swapaxes

```python
var9 = np.array([[1,2,3],[4,5,6]])
print(var9)
print()
print(np.swapaxes(var9,0,1))
```

```python
v = np.matrix([[1,2],[3,4]])
print(v)
print()
print(np.swapaxes(v,0,1))
```

# Inverse

```python
n = np.matrix([[1,2],[3,4]])
print(n)
print()
print(np.linalg.inv(n))
```

# Power

```python
n1 = np.matrix([[1,2],[3,4]])
print(n1)
print()
print(np.linalg.matrix_power(n1,4))
```

```python
n1 = np.matrix([[1,2],[3,4]])
print(n1)
print()
print(np.linalg.matrix_power(n1,0))
```

```python
n1 = np.matrix([[1,2],[3,4]])
print(n1)
print()
print(np.linalg.matrix_power(n1,-2))
```

# Determinate

```python
v = np.matrix([[1,2],[3,4]])
print(v)
print()
print(np.linalg.det(v))
```

```python
v = np.matrix([[1,2,3],[3,4,3],[1,2,3]])
print(v)
print()
print(np.linalg.det(v))
```
