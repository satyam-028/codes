# Insert

```python
import numpy as np

var = np.array([1,2,3,4])
print(var)
print(type(var))
print(var.dtype)
```

```python
var = np.array([1,2,3,4])
print(var)
c = np.insert(var,3,50)
print(c)
```

```python
var = np.array([1,2,3,4])
print(var)
c = np.insert(var,(2,4),50)
print(c)
```

```python
var = np.array([1,2,3,4])
print(var)
c = np.insert(var,(2,4),50.3)
print(c)
```

```python
v = np.array([[1,2,3],[5,6,7]])
s = np.insert(v,2,6,axis=0)
print(s)
```

```python
v = np.array([[1,2,3],[5,6,7]])
s = np.insert(v,2,6,axis=1)
print(s)
```

```python
v = np.array([[1,2,3],[5,6,7]])
s = np.insert(v,2,[22,23],axis=1)
print(s)
```

```python
v = np.array([1,2,3,5,6,7])
x = np.append(v,200)
print(x)
```

```python
v = np.array([[1,2,3],[5,6,7]])
s = np.append(v,[[22,23,33]],axis=0)
print(s)
```

# Delete

```python
v = np.array([1,2,3,5,6,7])
x = np.delete(v,2)
print(x)
```
