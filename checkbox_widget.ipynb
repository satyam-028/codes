# using variable

```python
from tkinter import *

win = Tk()

var = BooleanVar()
cb = Checkbutton(win, text="python",font=(30),variable=var)
cb.pack()

print(var.get())

win.mainloop()
```

# using button in variable

```python
from tkinter import *
def test():
    print(var.get())
    

win = Tk()

var = BooleanVar()
cb = Checkbutton(win, text="python",font=(30),variable=var)
cb.pack()

bt = Button(win, text="python",font=(30), command=test)
bt.pack()
print(var.get())

win.mainloop()
```

```python
from tkinter import *
def test():
    lb.config(text=var.get())

win = Tk()

var = StringVar()
cb = Checkbutton(win, text="python",font=(30),onvalue="python",offvalue="java",variable=var)
cb.pack()

lb = Label(win, text="", font=(30))
lb.pack()

bt = Button(win, text="python",font=(30), command=test)
bt.pack()
print(var.get())

win.mainloop()
```

# deselect() parameter

```python
from tkinter import *
def test():
    lb.config(text=var.get())

win = Tk()

var = StringVar()

cb = Checkbutton(win, text="python",font=(30),onvalue="python",offvalue="java",variable=var)
cb.deselect() # tick mark are remove with this parameter
cb.pack()

lb = Label(win, text="", font=(30))
lb.pack()

bt = Button(win, text="python",font=(30), command=test)
bt.pack()
print(var.get())

win.mainloop()
```
