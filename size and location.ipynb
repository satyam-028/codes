# window in center 

```python
from tkinter import *


win = Tk()

width = 300
height = 500

# win['bg']="black" ## change background color 

sys_width = win.winfo_screenwidth()
sys_height = win.winfo_screenheight()

c_x = int(sys_width/2 - width/2)
c_y = int(sys_height/2 - height/2)

win.geometry(f"{width}x{height}+{c_x}+{c_y}")  ## size of the window and location of the window

win['bg']="black" ## change background color 

win.mainloop()
```

# minsize and maxsize of window

```python
from tkinter import *


win = Tk()

width = 300
height = 500

win.minsize(100,100)
win.maxsize(500,500)

win['bg']="black" ## change background color 

win.mainloop()
```

#  resizable parameter

```python
from tkinter import *


win = Tk()


win.title("***  Satyam Sharma  ***") ## give a title on top 

win.attributes('-alpha',1) ## transparent the window

# win.config(bg="red") ## change background color 
win['bg']="black" ## change background color 

win.geometry("300x500-50-50")  ## size of the window and location of the window

win.resizable(False,False)

win.mainloop()
```
